```python
def is_safe(board, row, col, n):
    for i in range(n):
        if board[i][col] == 1:
            return False

    i, j = row - 1, col - 1
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    i, j = row - 1, col + 1
    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    i, j = row + 1, col - 1
    while i < n and j >= 0:
        if board[i][j] == 1:
            return False
        i += 1
        j -= 1

    i, j = row + 1, col + 1
    while i < n and j < n:
        if board[i][j] == 1:
            return False
        i += 1
        j += 1

    return True

def solve_n_queens(board, row, n, fixed_row):
    if row == n:
        return True

    if row == fixed_row:
        return solve_n_queens(board, row + 1, n, fixed_row)

    for col in range(n):
        if is_safe(board, row, col, n):
            board[row][col] = 1

            if solve_n_queens(board, row + 1, n, fixed_row):
                return True

            board[row][col] = 0

    return False

n = int(input("Enter value of N: "))

if n <= 0:
    print("N must be a positive number.")
else:
    first_row = int(input("Enter first queen row (0 to N-1): "))
    first_col = int(input("Enter first queen column (0 to N-1): "))

    if not (0 <= first_row < n and 0 <= first_col < n):
        print("Invalid position.")
    else:
        board = [[0 for j in range(n)] for i in range(n)]
        board[first_row][first_col] = 1

        if solve_n_queens(board, 0, n, first_row):
            print("\\nN-Queens solution (1 = Queen, 0 = Empty):")
            for row in board:
                print(row)
        else:
            print("No solution exists for this first-queen position.")
```
