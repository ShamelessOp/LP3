```python
def fractional_knapsack(weights, values, capacity):
    items = []

    for i in range(len(weights)):
        ratio = values[i] / weights[i]
        items.append((ratio, weights[i], values[i], i + 1))

    items.sort(reverse=True)

    total_value = 0.0
    selected = []

    for ratio, weight, value, item_no in items:
        if capacity == 0:
            break

        if weight <= capacity:
            capacity -= weight
            total_value += value
            selected.append((item_no, 1.0, value))
        else:
            fraction = capacity / weight
            total_value += value * fraction
            selected.append((item_no, fraction, value * fraction))
            capacity = 0

    return total_value, selected

n = int(input("Enter number of items: "))
weights = []
values = []

for i in range(n):
    print("\\nItem", i + 1)
    weights.append(float(input("Enter weight: ")))
    values.append(float(input("Enter value: ")))

capacity = float(input("\\nEnter knapsack capacity: "))

if n < 0 or capacity < 0 or any(w <= 0 for w in weights):
    print("Enter valid positive item weights and non-negative capacity.")
else:
    max_value, selected_items = fractional_knapsack(weights, values, capacity)
    print("\\nItems taken (item number, fraction, value added):")
    for item in selected_items:
        print(item)
    print("Maximum value:", round(max_value, 2))
```
