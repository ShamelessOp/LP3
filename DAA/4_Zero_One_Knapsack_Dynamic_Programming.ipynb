```python
def knapsack_01(weights, values, capacity):
    n = len(weights)

    dp = [[0 for w in range(capacity + 1)] for i in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i - 1] <= w:
                take = values[i - 1] + dp[i - 1][w - weights[i - 1]]
                leave = dp[i - 1][w]
                dp[i][w] = max(take, leave)
            else:
                dp[i][w] = dp[i - 1][w]

    selected = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            selected.append(i)
            w -= weights[i - 1]

    selected.reverse()
    return dp[n][capacity], selected

n = int(input("Enter number of items: "))
weights = []
values = []

for i in range(n):
    print("\\nItem", i + 1)
    weights.append(int(input("Enter weight: ")))
    values.append(int(input("Enter value: ")))

capacity = int(input("\\nEnter knapsack capacity: "))

if n < 0 or capacity < 0 or any(w < 0 for w in weights):
    print("Please enter valid non-negative values.")
else:
    maximum, selected = knapsack_01(weights, values, capacity)
    print("\\nMaximum value:", maximum)
    print("Selected item numbers:", selected)
```
